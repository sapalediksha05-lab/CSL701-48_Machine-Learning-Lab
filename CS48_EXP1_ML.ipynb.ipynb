Task 7: Generate summary statistics using df.describe(); record mean, minimum and maximum values of Age and Fare.


```python
description = df.describe()
display(description)
```

Display mean, minimum, and maximum values of Age and Fare

```python
print("\nAge Statistics:")
print("Mean Age:", df["Age"].mean())
print("Minimum Age:", df["Age"].min())
print("Maximum Age:", df["Age"].max())

print("\nFare Statistics:")
print("Mean Fare:", df["Fare"].mean())
print("Minimum Fare:", df["Fare"].min())
print("Maximum Fare:", df["Fare"].max())
```

Task 8: Identify missing values

```python
missing_values = df.isnull().sum()
display(missing_values)
```

Task 9: Fill missing Age values with the median

```python
df["Age"].fillna(df["Age"].median(), inplace=True)
```

```python
# Check if missing values in Age are filled
print("Missing values in Age after filling:")
print(df["Age"].isnull().sum())
```

Task 10: Filter surviving passengers (Survived == 1)

```python
survived_df = df[df['Survived'] == 1]
display(survived_df.head())
```

Task 11: Convert Fare column to a NumPy array.



```python
fare_array = df['Fare'].to_numpy()
print(f"Type of fare_array: {type(fare_array)}")
display(fare_array[:5]) # Display first 5 elements for verification
```

Task 12: Calculate maximum and average Fare using np.max() and np.mean().


```python
max_fare = np.max(fare_array)
mean_fare = np.mean(fare_array)
print(f"Maximum Fare: {max_fare:.2f}")
print(f"Average Fare: {mean_fare:.2f}")
```

Task 13: Plot histogram of Age.


```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

Task 14: Plot bar chart of Survival Count.


```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Survived', data=df, palette='viridis')
plt.title('Survival Count (0 = Not Survived, 1 = Survived)')
plt.xlabel('Survival Status')
plt.ylabel('Number of Passengers')
plt.xticks(ticks=[0, 1], labels=['Not Survived', 'Survived'])
plt.grid(axis='y', alpha=0.75)
plt.show()
```

Task 15: Plot scatter graph of Age vs Fare

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='Fare', data=df, hue='Survived', palette='viridis', alpha=0.7)
plt.title('Scatter Plot of Age vs. Fare (Colored by Survival)')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Survived', loc='upper right', labels=['Not Survived', 'Survived'])
plt.show()
```
